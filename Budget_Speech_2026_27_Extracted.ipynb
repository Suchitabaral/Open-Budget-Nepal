```python
import pdfplumber
from pathlib import Path

# Folder path
folder_path = Path.home() / "Desktop" / "2083_84"

# PDF file
pdf_file = folder_path / "Budget Speech 2026_27 English Translation_ysmw9xe.pdf"

# Output text file
output_file = folder_path / "Budget_Speech_2026_27_Extracted.txt"

all_text = ""

# Extract text from PDF
with pdfplumber.open(pdf_file) as pdf:
    for page_num, page in enumerate(pdf.pages, start=1):
        text = page.extract_text()
        
        if text:
            all_text += f"\n\n--- Page {page_num} ---\n\n"
            all_text += text

# Save extracted text
with open(output_file, "w", encoding="utf-8") as f:
    f.write(all_text)

print(f"Text extracted successfully!")
print(f"Saved to: {output_file}")
```
