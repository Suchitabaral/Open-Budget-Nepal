```python
import pdfplumber
import pandas as pd
from pathlib import Path

folder_path = Path.home() / "Desktop" / "2083_84"
pdf_file = folder_path / "व्यय अनुमानको विवरण (रातो किताब ) आ.व. २०८३-८४_c03uuu0.pdf"

output_folder = folder_path / "extracted_tables"
output_folder.mkdir(exist_ok=True)

table_count = 0

with pdfplumber.open(pdf_file) as pdf:
    for page_num, page in enumerate(pdf.pages, start=1):
        tables = page.extract_tables()

        for table_num, table in enumerate(tables, start=1):
            if table:
                df = pd.DataFrame(table)

                csv_file = output_folder / f"page_{page_num}_table_{table_num}.csv"
                df.to_csv(csv_file, index=False, encoding="utf-8-sig")

                table_count += 1

print(f"Saved {table_count} tables to {output_folder}")
```
